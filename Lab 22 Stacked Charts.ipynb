### Task 1: Stacked Chart for Composition of Job Satisfaction Across Age Groups


##### 1. Stacked Chart of Median `JobSatPoints_6` and `JobSatPoints_7` for Different Age Groups


Visualize the composition of job satisfaction scores (`JobSatPoints_6` and `JobSatPoints_7`) across various age groups. This will help in understanding the breakdown of satisfaction levels across different demographics.



```python
age_sat = df.groupby('Age')[['JobSatPoints_6', 'JobSatPoints_7']].median()

age_sat.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.xticks(rotation=45, ha='right')
plt.xlabel('Age Group')
plt.ylabel('Median Job Satisfaction')
plt.title('Median Job Satisfaction by Age Group')
plt.tight_layout()
plt.show()
```

##### Stacked Chart of `JobSatPoints_6` and `JobSatPoints_7` for Employment Status


Create a stacked chart to compare job satisfaction (`JobSatPoints_6` and `JobSatPoints_7`) across different employment statuses. This will show how satisfaction varies by employment type.


```python
top5_emp = df['Employment'].value_counts().head(5).index
df_emp = df[df['Employment'].isin(top5_emp)]

emp_sat = df_emp.groupby('Employment')[['JobSatPoints_6', 'JobSatPoints_7']].median()

emp_sat.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.xticks(rotation=45, ha='right')
plt.xlabel('Employment Type')
plt.ylabel('Median Job Satisfaction')
plt.title('Median Job Satisfaction by Employment Type')
plt.tight_layout()
plt.show()
```

### Task 2: Stacked Chart for Compensation and Job Satisfaction by Age Group


##### This stacked chart visualizes the composition of compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`) specifically for respondents aged 30-35.


```python
df25to34 = df[df['Age'] == '25-34 years old']

cap = df25to34['ConvertedCompYearly'].quantile(0.95)
df25to34 = df25to34[df25to34['ConvertedCompYearly'] <= cap]

stacked = df25to34[['ConvertedCompYearly', 'JobSatPoints_6']].median().to_frame().T
stacked.plot(kind='bar', stacked=True, figsize=(8, 6))
plt.title('Median Compensation and Job Satisfaction (Age 25-34)')
plt.ylabel('Median Value')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

##### Stacked Chart of Median Compensation and Job Satisfaction Across Age Group


Compare the median compensation and job satisfaction metrics across different age groups. This helps visualize how compensation and satisfaction levels differ by age.


```python
cap = df['ConvertedCompYearly'].quantile(0.95)
df_capped = df[df['ConvertedCompYearly'] <= cap]

age_comp_sat = df_capped.groupby('Age')[['ConvertedCompYearly', 'JobSatPoints_6']].median()
age_comp_sat.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.xticks(rotation=45, ha='right')
plt.xlabel('Age Group')
plt.ylabel('Median Value')
plt.title('Median Compensation and Job Satisfaction by Age Group')
plt.tight_layout()
plt.show()
```

### Task 3: Comparing Data Using Stacked Charts


##### 1. Stacked Chart of Preferred Databases by Age Group




Visualize the top databases that respondents from different age groups wish to learn. Create a stacked chart to show the proportion of each database in each age group.


```python
df_db = df.assign(Database=df['DatabaseWantToWorkWith'].str.split(';')).explode('Database').reset_index(drop=True)

top5_db = df_db['Database'].value_counts().head(5).index
df_db_filtered = df_db[df_db['Database'].isin(top5_db)]

ct = pd.crosstab(df_db_filtered['Age'], df_db_filtered['Database'])
ct.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.xticks(rotation=45, ha='right')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.title('Preferred Databases by Age Group')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=7)
plt.tight_layout()
plt.show()
```

##### 2. Stacked Chart of Employment Type by Job Satisfaction


Analyze the distribution of employment types within each job satisfaction level using a stacked chart. This will provide insights into how employment types are distributed across various satisfaction ratings.


```python
top5_emp = df['Employment'].value_counts().head(5).index
df_emp = df[df['Employment'].isin(top5_emp)].dropna(subset=['JobSat'])

df_emp['JobSat'] = df_emp['JobSat'].astype(int)

ct2 = pd.crosstab(df_emp['JobSat'], df_emp['Employment'])
ct2.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.xticks(rotation=0)
plt.xlabel('Job Satisfaction')
plt.ylabel('Count')
plt.title('Employment Type by Job Satisfaction Level')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=7)
plt.tight_layout()
plt.show()
```

### Task 4: Exploring Technology Preferences Using Stacked Charts


##### 1. Stacked Chart for Preferred Programming Languages by Age Group


Analyze how programming language preferences (`LanguageAdmired`) vary across age groups.


```python
df_lang = df.assign(Language=df['LanguageAdmired'].str.split(';')).explode('Language').reset_index(drop=True)

top5_langs = df_lang['Language'].value_counts().head(5).index
df_lang_filtered = df_lang[df_lang['Language'].isin(top5_langs)]

ct = pd.crosstab(df_lang_filtered['Age'], df_lang_filtered['Language'])
ct.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.xticks(rotation=45, ha='right')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.title('Admired Programming Languages by Age Group')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=7)
plt.tight_layout()
plt.show()
```

##### 2. Stacked Chart for Technology Adoption by Employment Type


Explore how admired platforms (`PlatformAdmired`) differ across employment types (e.g., full-time, freelance)


```python
df_plat = df.assign(Platform=df['PlatformAdmired'].str.split(';')).explode('Platform').reset_index(drop=True)

top5_platforms = df_plat['Platform'].value_counts().head(5).index
top5_emp = df['Employment'].value_counts().head(5).index
df_plat_filtered = df_plat[df_plat['Platform'].isin(top5_platforms) & df_plat['Employment'].isin(top5_emp)]

ct2 = pd.crosstab(df_plat_filtered['Employment'], df_plat_filtered['Platform'])
ct2.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.xticks(rotation=45, ha='right')
plt.xlabel('Employment Type')
plt.ylabel('Count')
plt.title('Admired Platforms by Employment Type')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=7)
plt.tight_layout()
plt.show()
```
